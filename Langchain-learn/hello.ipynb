```python
from langchain.tools import tool
import json
import urllib.parse
import urllib.request


@tool
def get_weather(city: str) -> str:
    """???????????????????????????Shenzhen?"""
    city = city.strip()
    if not city:
        return "????????"

    query = urllib.parse.quote(city)
    url = f"https://wttr.in/{query}?format=j1&lang=zh"

    try:
        request = urllib.request.Request(
            url,
            headers={"User-Agent": "langchain-weather-tool/1.0"},
        )
        with urllib.request.urlopen(request, timeout=10) as response:
            data = json.loads(response.read().decode("utf-8"))
    except Exception as exc:
        return f"???????{exc}"

    current = data["current_condition"][0]
    area = data.get("nearest_area", [{}])[0]
    location = area.get("areaName", [{"value": city}])[0]["value"]
    description = current.get("lang_zh", current.get("weatherDesc", [{"value": "??"}]))[0]["value"]

    return (
        f"{location} ?????{description}?"
        f"?? {current['temp_C']}?C??? {current['FeelsLikeC']}?C?"
        f"?? {current['humidity']}%??? {current['windspeedKmph']} km/h?"
    )
```
