工具

```python
from langchain.tools import tool
import json
import urllib.parse
import urllib.request


@tool
def get_weather(city: str) -> str:
    """
    查询指定城市的天气情况。

    当用户询问天气相关问题时必须调用本工具，
    不允许自行编造天气信息。

    支持问题示例：
    - 北京天气怎么样
    - 上海今天热不热
    - 深圳是否下雨

    Args:
        city: 城市名称
    """
    city = city.strip()
    return "当前城市"+city+"天气阴"
```

创建agent

```python
from langchain_openai import ChatOpenAI
import os
from langchain.agents import create_agent


model = ChatOpenAI(
    model="qwen3.5-plus",
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url=os.getenv("OPENAI_BASE_URL"),
)

agent=create_agent(
    model=model,
    tools=[get_weather]
)
```

发起对模型的调用

```python
res=await agent.ainvoke({
    "messages":[
        {"role":"system","content":"你是一个天气助手。遇到天气问题必须调用 get_weather 工具不允许自己回答。"},
        {"role":"user","content":"今天成都天气怎么样"}
    ]
})
```
